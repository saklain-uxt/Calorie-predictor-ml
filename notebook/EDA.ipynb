### 3.7 Exploring Data

```python
print("Categories in 'gender' variable:     ",end=" " )
print(df['Gender'].unique())

```

```python
# define numerical & categorical columns

numeric_features = [feature for feature in df.columns if df[feature].dtype != 'str']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'str']

print(' numerical features: ', numeric_features, '\nNumber of numerical features: ', len(numeric_features))
print(' categorical features: ', categorical_features, '\nNumber of categorical features: ', len(categorical_features))

print("we have {} numerical feature and {} categorical feature ".format(len(numeric_features), len(categorical_features)))
```

```python
df.head(2)
```

```python
sns.countplot(data=df, x='Gender')
plt.title('Gender Distribution')
plt.show()
```

```python
df.hist(figsize=(12, 10))
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df, x='Duration', y='Calories')
plt.title('Duration vs Calories')
plt.show()
```

```python
sns.scatterplot(data=df, x='Weight', y='Calories')
plt.title('Weight vs Calories')
plt.show()
```

```python
sns.scatterplot(data=df, x='Age', y='Calories')
plt.title('Age vs Calories')
plt.show()
```

```python
sns.scatterplot(data=df, x='Body_Temp', y='Calories')
plt.title('Body Temperature vs Calories')
plt.show()
```

```python
correlation = df.select_dtypes(include='number').corr()

plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.boxplot(data=df, x='Calories')
plt.title('Calories Outliers')
plt.show()
```

```python
sns.boxplot(data=df[['Age', 'Height', 'Weight', 'Duration',
                     'Heart_Rate', 'Body_Temp', 'Calories']])
plt.xticks(rotation=45)
plt.show()
```
